```python
import pandas as pd
import os
os.chdir('D:/Projects/tumor-immune-scrna-atlas')
print(os.getcwd())
# Load just LJQ to confirm format
df = pd.read_csv('data/raw/GSM5219674_LJQ-T.counts.tsv.gz',
                 sep='\t', index_col=0)
print('Shape (genes × cells):', df.shape)
print('First 3 genes:', df.index[:3].tolist())
print('First 3 cells:', df.columns[:3].tolist())
print('Max count:', df.values.max())   # should be integer-like
print('Memory (dense):', df.memory_usage(deep=True).sum() / 1e6, 'MB')
```
